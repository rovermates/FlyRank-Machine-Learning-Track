## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: given limited editor time, which pages should be reviewed for refresh first?

Who acts: a content editor or SEO reviewer with a fixed weekly capacity (e.g., can only review ~50 pages).

The action: the editor opens the top-ranked pages and decides whether to update, expand, consolidate, or leave them.
Cost of a false positive: the editor spends limited review time on a page that was not actually a problem, wasted hours, and a genuinely declining page waits longer.
Cost of a false negative: a page that iss actually losing visibility never surfaces on the list, so nobody reviews it until the loss is worse.
Why data/ML helps at all: with tens of thousands of pages and a handful of reviewer-hours, a hand-picked rule (stale AND visible) is cheap but crude, it can not weigh multiple weak signals (position, CTR, freshness, word count) against each other the way a trained ranker can.
That tradeoff is worth testing, not assuming.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

These numbers say two things: first, a meaningful slice of pages qualify as "stale and visible" — enough volume that a ranking method has something real to work with.
```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/rovermates/FlyRank-Machine-Learning-Track/"
REPO_DIR = "FlyRank-Machine-Learning-Track"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("../..")

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_rate = df["trend_direction"].str.lower().eq("down").mean()
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
total_pages = len(df)

print(f"Total pages: {total_pages}")
print(f"Declining rate (trend_direction == down): {declining_rate:.1%}")
print(f"Pages that are stale AND still visible (candidates for review): {stale_visible} ({stale_visible/total_pages:.1%} of all pages)")
print()
print("From the verified starter pipeline (outputs/model_results.json):")
print("  Baseline rule   Precision@50 = 0.240  (~12 of top 50 correct)")
print("  Random forest   Precision@50 = 0.740  (~37 of top 50 correct)")
```
